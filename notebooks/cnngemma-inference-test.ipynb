```python
import sys
import os
sys.path.append(os.path.abspath("../src"))
from utils import load_finetuned_model
import torch

from inference import inference
from cnn_gemma import CNNGemmaConfig
from processing_cnngemma import CNNGemmaProcessor
import json

image_file_path = "..\\dataset\\RISCM\\resized\\NWPU_361.jpg"
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python

with open("../src/configs/mobilenet_single.json", "r") as f:
    model_config_file = json.load(f)

config = CNNGemmaConfig(**model_config_file)
model, tokenizer = load_finetuned_model(model_path="../src/weights/mobilenet_single",hub_id="ahmetbekcan/CNNGemma-MobileNet-Single-224",device="cuda", config=config, dtype=torch.bfloat16)
processor = CNNGemmaProcessor(tokenizer=tokenizer,num_image_tokens=model.config.vision_config.num_image_tokens, image_encoder_type=model.config.vision_config.architecture)

prediction = inference(model, processor, prompt="caption en", image_file_path=image_file_path, device=device, do_sample=False)
print(prediction)
```
