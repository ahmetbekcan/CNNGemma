```python
import os
import sys
sys.path.append(os.path.abspath("../src"))
import pandas as pd


def averages_of_max_metrics(csv_path):
    df = pd.read_csv(csv_path)
    
    metric_columns = ['bleu_max', 'meteor_max', 'rouge_max']
    
    averages = df[metric_columns].mean()
    return averages
```

```python
mobile_single_path = "../src/evaluate/mobilenet_single/evaluation_results.csv"
mobile_mutliple_path = "../src/evaluate/mobilenet_multiple/evaluation_results.csv"
efficient_single_path = "../src/evaluate/efficientnet_single/evaluation_results.csv"
efficient_multiple_path = "../src/evaluate/efficientnet_multiple/evaluation_results.csv"
paligemma_pretrained_path = "../src/evaluate/paligemma/evaluation_results.csv"
paths = {
    "MobileNet-Single": mobile_single_path,
    "MobileNet-Multiple": mobile_mutliple_path,
    "EfficientNet-Single": efficient_single_path,
    "EfficientNet-Multiple": efficient_multiple_path,
    "PaliGemma-Pretrained": paligemma_pretrained_path
}
```

```python
for name, path in paths.items():
    if os.path.exists(path):
        avg_max = averages_of_max_metrics(path)
        print(f"Results of {name}:")
        for metric, value in avg_max.items():
            print(f"Average of {metric}: {value:.4f}")
        print(f"Average of all metrics: {avg_max.mean()}")
    else:
        print(f"{name}: File not found at {path}")
    print("\n")
```
